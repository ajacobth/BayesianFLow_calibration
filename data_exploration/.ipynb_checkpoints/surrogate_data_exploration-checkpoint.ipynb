## Train 


```python


# ---- load ----
df = pd.read_csv("Train_Set_2.csv")#("Completed_train_set_10_30_25.csv")

# first 5 = inputs, last = output
X = df.iloc[:, :5].astype(float)
y = df.iloc[:, -1].astype(float)

# ---- standardize (z-score, ddof=0 to match numpy) ----
Xz = (X - X.mean()) / X.std(ddof=0)
yz = (y - y.mean()) / y.std(ddof=0)

# ---- 1) trace plots of standardized columns (inputs + output) ----
for c in Xz.columns:
    plt.figure(figsize=(7, 3))
    plt.plot(Xz[c].values)
    plt.title(f"Trace (standardized) — {c}")
    plt.xlabel("Row index"); plt.ylabel("z-score")
    plt.tight_layout()
    plt.show()

plt.figure(figsize=(7, 3))
plt.plot(yz.values)
plt.title("Trace (standardized) — output")
plt.xlabel("Row index"); plt.ylabel("z-score")
plt.tight_layout(); plt.show()

# ---- 2) scatter: each input vs output (standardized) ----
for c in Xz.columns:
    plt.figure(figsize=(5.5, 4.5))
    plt.scatter(Xz[c].values, yz.values, alpha=0.6)
    plt.xlabel(f"{c} (z)"); plt.ylabel("output (z)")
    plt.title(f"{c} vs output (standardized)")
    plt.tight_layout()
    plt.show()

# ---- 3) seaborn pairplot for input space (standardized) ----
sns.pairplot(Xz, corner=True, diag_kind="kde")
plt.suptitle("Inputs pairplot (standardized)", y=1.02)
plt.show()


```

# Marginalize over other varobles to understand effect of each

```python
import numpy as np
import matplotlib.pyplot as plt

features = Xz.columns[:5]         # x1..x5
n_bins = 20

fig, axes = plt.subplots(3, 2, figsize=(10, 9), sharey=True)
axes = axes.ravel()

for k, col in enumerate(features):
    ax = axes[k]
    # bin xi
    bins = np.linspace(Xz[col].min(), Xz[col].max(), n_bins)
    idx = np.digitize(Xz[col].values, bins)
    # mean(y) per bin (ignore empty-bin warnings -> NaNs)
    means = [np.nanmean(yz[idx == i]) for i in range(1, len(bins))]
    # plot at bin centers
    centers = 0.5 * (bins[:-1] + bins[1:])
    ax.plot(centers, means, marker='o')
    ax.set_xlabel(f"{col} (z)")
    ax.set_ylabel("Mean output (z)")
    ax.set_title(f"Main effect of {col}")

# hide the unused 6th subplot
for j in range(len(features), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

## Test


```python


# ---- load ----
df = pd.read_csv("Completed_test_set_10_30_25.csv")

# first 5 = inputs, last = output
X = df.iloc[:, :5].astype(float)
y = df.iloc[:, -1].astype(float)

# ---- standardize (z-score, ddof=0 to match numpy) ----
Xz = (X - X.mean()) / X.std(ddof=0)
yz = (y - y.mean()) / y.std(ddof=0)

# ---- 1) trace plots of standardized columns (inputs + output) ----
for c in Xz.columns:
    plt.figure(figsize=(7, 3))
    plt.plot(Xz[c].values)
    plt.title(f"Trace (standardized) — {c}")
    plt.xlabel("Row index"); plt.ylabel("z-score")
    plt.tight_layout()
    plt.show()

plt.figure(figsize=(7, 3))
plt.plot(yz.values)
plt.title("Trace (standardized) — output")
plt.xlabel("Row index"); plt.ylabel("z-score")
plt.tight_layout(); plt.show()

# ---- 2) scatter: each input vs output (standardized) ----
for c in Xz.columns:
    plt.figure(figsize=(5.5, 4.5))
    plt.scatter(Xz[c].values, yz.values, alpha=0.6)
    plt.xlabel(f"{c} (z)"); plt.ylabel("output (z)")
    plt.title(f"{c} vs output (standardized)")
    plt.tight_layout()
    plt.show()

# ---- 3) seaborn pairplot for input space (standardized) ----
sns.pairplot(Xz, corner=True, diag_kind="kde")
plt.suptitle("Inputs pairplot (standardized)", y=1.02)
plt.show()
```

```python
mask = Xz["x1"] > 1.25
indices = np.where(mask)[0]

# plot full trace
plt.figure(figsize=(10, 3.5))
plt.plot(yz, label="output (z)", color="steelblue")

# overlay highlighted points
plt.scatter(indices, yz[mask], color="crimson", s=30, label="x1 > 1.25", zorder=3)

plt.xlabel("Row index")
plt.ylabel("output (z)")
plt.title("Trace (standardized) — output\nhighlighted where x1 > 1.25")
plt.legend()
plt.tight_layout()
plt.show()
```
